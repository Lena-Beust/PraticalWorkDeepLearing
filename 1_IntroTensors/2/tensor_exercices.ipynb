## Exercise 3 — Reshaping tensors

Reshaping changes how elements are arranged into dimensions, without
changing the elements themselves.  The total number of elements must stay
the same.

### 3.1  Reshape a 1-D tensor into a 2-D matrix

Take the tensor `[0, 1, 2, ..., 11]` and reshape it into a matrix with

Reshaped shape: torch.Size([3, 4])

````python
flat = torch.arange(12)
print("Original shape:", flat.shape)

# === YOUR CODE ===
reshaped = flat.reshape([3,4])
            # or torch.reshape(flat,(3,4))
# =================

print("Reshaped:")
print(reshaped)
print("Reshaped shape:", reshaped.shape)

print("\n--- Expected ---")
print("Original shape: torch.Size([12])")
print("Reshaped:")
print("tensor([[ 0,  1,  2,  3],")
print("        [ 4,  5,  6,  7],")
print("        [ 8,  9, 10, 11]])")
print("Reshaped shape: torch.Size([3, 4])")
````

## Exercise 6 — Image-shaped tensors

In deep learning for computer vision, images are represented as tensors
with specific dimension conventions:
- A single greyscale image: `(1, H, W)`
- A single RGB image: `(3, H, W)`
- A batch of images: `(N, C, H, W)`

where `N` = batch size, `C` = channels, `H` = height, `W` = width.

### 6.2  Create a fake RGB image and reorder its channels

1. Create a random tensor of shape `(32, 32, 3)` — this is the NumPy/PIL
   channel-last convention `(H, W, C)`.
2. Convert it to PyTorch channel-first format `(C, H, W)`.

Expected output:
```
Before permute (H,W,C): torch.Size([32, 32, 3])
After  permute (C,H,W): torch.Size([3, 32, 32])
```

````python
# === YOUR CODE ===
img_hwc = torch.rand(32, 32, 3) # Random tensor of shape (32, 32, 3)
img_chw = img_hwc.permute(2, 0, 1) # Permute img_hwc to shape (3, 32, 32)
# =================

print("Before permute (H,W,C):", img_hwc.shape)
print("After  permute (C,H,W):", img_chw.shape)

print("\n--- Expected ---")
print("Before permute (H,W,C): torch.Size([32, 32, 3])")
print("After  permute (C,H,W): torch.Size([3, 32, 32])")
````

### 6.3  Add a batch dimension with `unsqueeze`

Take the channel-first image `img_chw` (shape `(3, 32, 32)`) and add a
batch dimension at position 0, so the result has shape `(1, 3, 32, 32)`.

This is what you do when you want to feed a **single image** to a model
that expects a batch.

Expected output:
```
Single image:  torch.Size([3, 32, 32])
With batch dim: torch.Size([1, 3, 32, 32])
```

````python
# === YOUR CODE ===
batched = img_chw.unsqueeze(0) # Add a dimension at position 0 to img_chw
        # or img_chw.reshape(1,*img_chw.shape)
# =================

print("Single image: ", img_chw.shape)
print("With batch dim:", batched.shape)

print("\n--- Expected ---")
print("Single image:  torch.Size([3, 32, 32])")
print("With batch dim: torch.Size([1, 3, 32, 32])")
````

## Exercise 7 — Putting it all together

This exercise combines several operations into a small pipeline.

### 7.1  Mini pipeline

Starting from a NumPy array, perform the following steps:

1. Convert the NumPy array to a PyTorch float32 tensor.
2. Normalise the values to the range [0, 1].
3. Reshape from `(28, 28)` in order to add the channel dimension.
4. Add a batch dimension.

Expected output:
```
Step 0 — NumPy array:  shape=(28, 28), dtype=uint8
Step 1 — Tensor:       shape=torch.Size([28, 28]), dtype=torch.float32
Step 2 — Normalised:   min=0.0000, max=1.0000
Step 3 — With channel: torch.Size([1, 28, 28])
Step 4 — With batch:   torch.Size([1, 1, 28, 28])
```

````python
# Starting data: a fake 28x28 greyscale image as a uint8 NumPy array.
np_image = np.random.randint(0, 256, size=(28, 28), dtype=np.uint8)
print(f"Step 0 — NumPy array:  shape={np_image.shape}, dtype={np_image.dtype}")

# === YOUR CODE ===
# Step 1: Convert to a float32 PyTorch tensor
step1 = torch.from_numpy(np_image).float()
        # or torch.tensor(np_image, dtype=torch.float32)
# =================
print(f"Step 1 — Tensor:       shape={step1.shape}, dtype={step1.dtype}")

# === YOUR CODE ===
# Step 2: Normalise to [0, 1]
step2 = step1/step1.max()
# =================
print(f"Step 2 — Normalised:   min={step2.min():.4f}, max={step2.max():.4f}")

# === YOUR CODE ===
# Step 3: Add channel dimension -> (1, 28, 28)
step3 = step2.reshape(1, 28, 28)
# =================
print(f"Step 3 — With channel: {step3.shape}")

# === YOUR CODE ===
# Step 4: Add batch dimension -> (1, 1, 28, 28)
step4 = step3.unsqueeze(1)
        # or step3.reshape(1,1,28,28)
# =================
print(f"Step 4 — With batch:   {step4.shape}")

print("\n--- Expected ---")
print("Step 0 — NumPy array:  shape=(28, 28), dtype=uint8")
print("Step 1 — Tensor:       shape=torch.Size([28, 28]), dtype=torch.float32")
print("Step 2 — Normalised:   min=0.0000, max=1.0000")
print("Step 3 — With channel: torch.Size([1, 28, 28])")
print("Step 4 — With batch:   torch.Size([1, 1, 28, 28])")
````

<details>
<summary>Click for hints</summary>

- Step 1: `torch.from_numpy(...).float()` or `torch.tensor(..., dtype=torch.float32)`
- Step 2: `tensor / 255.0`
- Step 3: `.unsqueeze(0)` adds a dimension at position 0
- Step 4: `.unsqueeze(0)` again on the result of step 3

## Exercise 11 — DataLoader and batching

In practice, images are fed to models in **batches** via a `DataLoader`.
This adds a leading batch dimension.

### 11.1  Create DataLoaders and inspect a batch

1. Create a `DataLoader` for each training set.
2. Set a batch size of 32, with shuffled samples.
3. Fetch one batch.
4. Print the shape of the image batch and the label batch.

Expected output:
```
MNIST  batch images: torch.Size([32, 1, 28, 28])
MNIST  batch labels: torch.Size([32])
CIFAR  batch images: torch.Size([32, 3, 32, 32])
CIFAR  batch labels: torch.Size([32])
```

````python
# === YOUR CODE ===
mnist_loader = DataLoader(mnist_train, batch_size=32, shuffle=True)
cifar_loader = DataLoader(cifar_train, batch_size=32, shuffle=True)

mnist_batch_imgs, mnist_batch_labels = next(iter(mnist_loader))
cifar_batch_imgs, cifar_batch_labels = next(iter(cifar_loader))
# =================

print("MNIST  batch images:", mnist_batch_imgs.shape)
print("MNIST  batch labels:", mnist_batch_labels.shape)
print("CIFAR  batch images:", cifar_batch_imgs.shape)
print("CIFAR  batch labels:", cifar_batch_labels.shape)

print("\n--- Expected ---")
print("MNIST  batch images: torch.Size([32, 1, 28, 28])")
print("MNIST  batch labels: torch.Size([32])")
print("CIFAR  batch images: torch.Size([32, 3, 32, 32])")
print("CIFAR  batch labels: torch.Size([32])")
````

### 11.3  Compute per-channel pixel statistics

per channel over the batch.

channel dimension.
